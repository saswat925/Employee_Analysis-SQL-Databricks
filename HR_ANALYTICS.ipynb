```python
##5. Salary by Department (Bar Chart)
dept_salary = df.groupby("Department")["Monthly_Salary"].mean()

dept_salary.plot(kind="bar", figsize=(8,5))

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")

plt.xticks(rotation=45)

plt.show()
```

```python
##6. Performance by Department
performance = df.groupby("Department")["Performance_Score"].mean()

performance.plot(kind="bar", figsize=(8,5))

plt.title("Average Performance Score by Department")
plt.xlabel("Department")
plt.ylabel("Performance Score")

plt.xticks(rotation=45)

plt.show()
```

```python
##7. Satisfaction by Department
satisfaction = df.groupby("Department")["Employee_Satisfaction_Score"].mean()

satisfaction.plot(kind="bar", figsize=(8,5))

plt.title("Employee Satisfaction by Department")
plt.xlabel("Department")
plt.ylabel("Average Satisfaction")

plt.xticks(rotation=45)

plt.show()
```

```python
##8. Resigned vs Active
df["Resigned"].value_counts().plot(
    kind="bar",
    figsize=(6,5)
)

plt.title("Employee Attrition")
plt.xlabel("Resigned")
plt.ylabel("Employees")

plt.show()
```

```python
##9. Hiring Trend (Line Chart)
hire = df["Hire_Date"].dt.year.value_counts().sort_index()

hire.plot(figsize=(10,5), marker="o")

plt.title("Hiring Trend")
plt.xlabel("Year")
plt.ylabel("Employees Hired")

plt.show()
```

```python
##10. Overtime vs Performance
overtime = df.groupby("Overtime_Hours")["Performance_Score"].mean()

overtime.plot(figsize=(10,5))

plt.title("Overtime vs Performance")
plt.xlabel("Overtime Hours")
plt.ylabel("Average Performance")

plt.show()
```

```python
##11. Training vs Performance
training = df.groupby("Training_Hours")["Performance_Score"].mean()

training.plot(figsize=(10,5))

plt.title("Training Hours vs Performance")
plt.xlabel("Training Hours")
plt.ylabel("Average Performance")

plt.show()
```

```python
##CORRELATION HEATMAP
corr = df[
    [
        "Age",
        "Years_At_Company",
        "Performance_Score",
        "Monthly_Salary",
        "Work_Hours_Per_Week",
        "Projects_Handled",
        "Overtime_Hours",
        "Sick_Days",
        "Remote_Work_Frequency",
        "Team_Size",
        "Training_Hours",
        "Promotions",
        "Employee_Satisfaction_Score",
        "Resigned",
        "Hire_Year"
    ]
].corr()

print(corr)
```

```python
##visual correlation
corr = df.drop(columns=["Employee_ID"]).corr(numeric_only=True)

plt.figure(figsize=(10,8))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```

Employee_ID is a unique identifier, not a business metric. Including it in correlation analysis does not provide meaningful insights, so it was excluded before generating the correlation matrix.
